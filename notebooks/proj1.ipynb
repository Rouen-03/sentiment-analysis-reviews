```python
df = pd.read_csv(r'C:\Users\rouen\Downloads\customer_shopping_behavior.csv')
```

```python
df
```

```python
df.head()
```

```python
df['rev_rat_null_val'] = df.groupby('Category')['Review Rating'].transform('median')
```

```python
df.head()
```

```python
df['Review Rating'] = df['Review Rating'].fillna(df['rev_rat_null_val'])
```

```python
df.head()
```

```python
df.drop(columns = ['rev_rat_null_val'], inplace = True)
```

```python
df.head()
```

```python
df.isnull().sum()
```

```python
df.columns = df.columns.str.lower()
df.columns = df.columns.str.replace(' ','_')
```

```python
df.columns
```

```python
df = df.rename(columns = ({'purchase_amount_(usd)': 'purchase_amount'}))
```

```python
df['age_group'] = pd.cut(df['age'], 
       bins=[15, 25, 40, 60, 90],      # Range boundaries
       labels=['Young', 'Adult', 'Middle Age', 'Senior'],  # Group names
       right=False)                      # [20,30) means 20-29
```

```python
df[['age_group','age']].head(10)
```

```python
df.head()
```

```python
df['frequency_of_purchases'].unique()
```

```python
# Create a mapping dictionary
frequency_mapping = {
    'Annually': 365,
    'Quarterly': 90,
    'Every 3 Months': 90,
    'Monthly': 30,
    'Bi-Weekly': 14,
    'Fortnightly': 14,
    'Weekly': 7
}

# Apply mapping to create new column
df['purchase_frequency_days'] = df['frequency_of_purchases'].map(frequency_mapping)
```
