```python
import nltk

packages = [
    "stopwords", "punkt", "punkt_tab", "wordnet", "omw-1.4",
    "vader_lexicon", "averaged_perceptron_tagger", "averaged_perceptron_tagger_eng",
]

for pkg in packages:
    try:
        ok = nltk.download(pkg, quiet=True)
        print(pkg, "->", "ok" if ok else "failed")
    except Exception as e:
        print(pkg, "-> error:", e)
```
